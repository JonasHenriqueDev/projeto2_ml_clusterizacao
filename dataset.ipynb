```python
quantidade_nulos = df['abstract'].isnull().sum()
porcentagem_nulos = (df['abstract'].isnull().mean()) * 100

print(f"Quantidade de nulos em 'abstract': {quantidade_nulos}")
print(f"Porcentagem de nulos em 'abstract': {porcentagem_nulos:.2f}%")
```

Como o propósito do projeto é agrupar temas do RESUMO, então resumos nulos serão ignorados.

```python
#dropar linhas onde o resumo é nulo
df_clean = df.dropna(subset=['abstract'])

df_clean.head()
```

```python
quantidade_nulos = df_clean['abstract'].isnull().sum()
porcentagem_nulos = (df_clean['abstract'].isnull().mean()) * 100

print(f"Quantidade de nulos em 'abstract': {quantidade_nulos}")
print(f"Porcentagem de nulos em 'abstract': {porcentagem_nulos:.2f}%")
```

```python
df_clean.isnull().sum()
```

```python
#salvar dataframe limpo em formato parquet para melhor desempenho
df_clean.to_parquet('resources/dblp-v10-clean.parquet')
df_clean.to_csv('resources/dblp-v10-clean.csv')
```
