```python
import nltk
from nltk.corpus import stopwords
```

```python
nltk.download('stopwords')
```

```python
stopwords_en = set(stopwords.words('english'))
```

```python
print(f"Total de stopwords em inglês: {len(stopwords_en)}")

print(f"Lista das 10 primeiras stopwords: {list(stopwords_en)[:10]}")
```

Realizando tokenização e removendo stopwords na coluna abstract

```python
#remover stopwords
def remove_stopwords(text):
    # Verifica se o texto é uma string
    if isinstance(text, str):
        palavras = text.split()
        palavras_filtradas = [palavra for palavra in palavras if palavra.lower() not in stopwords_en]
        return ' '.join(palavras_filtradas)
    # Retorna o valor original se não for uma string (por exemplo, float)
    return text
```

```python
#Removendo pontos, vírgulas e outros caracteres especiais mantendo o hífen de palavras compostas
df['abstract'] = df['abstract'].str.replace(r'[^\w\s-]', '', regex=True)
```

```python
df['abstract'] = df['abstract'].apply(remove_stopwords)
```

```python
df['abstract'] = df['abstract'].str.lower()
```

```python
df['abstract'] = df['abstract'].str.split()
```
