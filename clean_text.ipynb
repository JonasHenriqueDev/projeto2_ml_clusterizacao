```python
import findspark
findspark.init()

from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("clusterizacao").config("spark.driver.memory", "10g").getOrCreate()
```

Limpando e normalizando o texto

```python
import nltk
from nltk.corpus import stopwords

from pyspark.ml.feature import StopWordsRemover, RegexTokenizer
from pyspark.sql.functions import col
```

```python
nltk.download('stopwords')
```

```python
stopwords_en = set(stopwords.words('english'))
```

```python
print(f"Total de stopwords em inglês: {len(stopwords_en)}")

print(f"Lista das 10 primeiras stopwords: {list(stopwords_en)[:10]}")
```

```python
import pandas as pd
```

```python
df = pd.read_csv('resources\dblp-v10-clean.csv')
```

```python
pd.set_option('display.max_colwidth', None)
df[['id', 'abstract']].head()
```

```python
df[['id', 'abstract']].tail()
```

Configurando o tokenizer

```python
tokenizer = RegexTokenizer(inputCol="abstract", outputCol="tokens", pattern="\\W")

df_spark = spark.createDataFrame(df)

#df_sample = df_spark.limit(5)

df_tokenizado = tokenizer.transform(df_spark)
```

```python
print("Dataframe tokenizado:")
df_tokenizado.select("id", "tokens").show(truncate=False)
```

```python
#Configurando o stopwords remover
stopwords_remover = StopWordsRemover(inputCol="tokens", outputCol="filtered", stopWords=list(stopwords_en))
```

```python
df_sem_stopwords = stopwords_remover.transform(df_tokenizado)
```

```python
df_sem_stopwords.select("id", "filtered").show(truncate=False)
```

Como podemos ver, o dataframe ainda contém tokens com apenas uma letra.
Iremos definir uma função para remover esses tokens

```python
print(df_sem_stopwords.columns)
```

```python
from pyspark.sql.functions import udf
from pyspark.sql.types import ArrayType, StringType
import re
```

```python
def remove_single_letter_tokens(tokens):
    return [token for token in tokens if len(token) > 1]

remove_single_letter_tokens_udf = udf(remove_single_letter_tokens, ArrayType(StringType()))

df_final = df_sem_stopwords.withColumn("filtered", remove_single_letter_tokens_udf(df_sem_stopwords["filtered"]))

df_final.show()
```

```python
df_final = df_final.withColumn("abstract", df_final["filtered"]) \
                   .drop("tokens", "filtered")
```

```python
df_final.show(truncate=False)
```

```python
df_final_reparticionado = df_final.repartition(20)
```

```python
spark.conf.set("spark.sql.execution.arrow.pyspark.enabled", "true")
```

```python
df_pandas = df_final_reparticionado.toPandas().
```

```python
df_final_reparticionado.write.parquet("resources/dblp_v10_clean-stopwords.parquet", mode="overwrite")
```
